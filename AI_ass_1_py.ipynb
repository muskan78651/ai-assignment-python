```python
# Question 6: Set Operations

list1 = list(map(int, input("Enter first list: ").split()))
list2 = list(map(int, input("Enter second list: ").split()))

set1 = set(list1)
set2 = set(list2)

print("Union:", set1 | set2)
print("Intersection:", set1 & set2)
print("Difference:", set1 - set2)
```

```python
# Question 9: Factorial

def factorial(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

num = int(input("Enter number: "))
print("Factorial:", factorial(num))
```

```python
# Question 10: Prime Checker

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

num = int(input("Enter number: "))

if is_prime(num):
    print("Prime")
else:
    print("Not Prime")
```

```python
# Question 11: Squares

def squares(lst):
    return [x**2 for x in lst]

numbers = list(map(int, input("Enter numbers: ").split()))
print(squares(numbers))
```

```python
# Question 13: Remove Duplicates

def remove_duplicates(lst):
    result = []
    for item in lst:
        if item not in result:
            result.append(item)
    return result

numbers = list(map(int, input("Enter numbers: ").split()))
print(remove_duplicates(numbers))
```

```python
# Question 14: Palindrome Checker

def is_palindrome(s):
    s = s.replace(" ", "").lower()
    return s == s[::-1]

text = input("Enter string: ")

if is_palindrome(text):
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Question 15: Fibonacci

def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        print(a, end=" ")
        a, b = b, a + b

n = int(input("Enter n: "))
fibonacci(n)
```

```python
# Question 16: Average Calculator

numbers = input("Enter numbers: ").split()

valid_numbers = []

for num in numbers:
    try:
        valid_numbers.append(float(num))
    except:
        print("Invalid input ignored:", num)

average = sum(valid_numbers) / len(valid_numbers)
print("Average:", average)
```

```python
# Question 19: Word Frequency

words = input("Enter words: ").split()

freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

print(freq)
```

```python
# Question 20: Temperature Converter

def convert():
    choice = input("Convert (C to F / F to C): ")

    if choice.lower() == "c to f":
        c = float(input("Enter Celsius: "))
        print("Fahrenheit:", (c * 9/5) + 32)

    elif choice.lower() == "f to c":
        f = float(input("Enter Fahrenheit: "))
        print("Celsius:", (f - 32) * 5/9)

    else:
        print("Invalid choice")

convert()
```
